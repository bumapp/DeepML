## Solution

### 1. Prove this expression:
$$ 
   \frac{d\sigma(x)}{dx} = \sigma(x)(1 - \sigma(x))
$$
#### Given
$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$
#### Step 1
$$
  \frac{d\sigma(x)}{dx} = -(1+e^{-x})^-2(-e^{-x})=\frac {e^{-x}}{(1 + e^{-x})^2}
$$
#### Step 2
$$
    \sigma(x)(1 - \sigma(x)) = (\frac{1}{1 + e^{-x}})(1-\frac{1}{1 + e^{-x}}) = (\frac{1}{1+e^{-x}})\cdot(\frac{e^{-x}}{1+e^{-x}}) = \frac {e^{-x}}{(1 + e^{-x})^2}
$$

#### Step 1 = Step 2 => The expression is proved
----------------------------------------------------------------------------------------------


### 2. Prove this expression:
$$
   \frac{d\tanh(x)}{dx} = 1 - \tanh(x)^2
$$

#### Given
$$
\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$

#### Step
$$
   \frac{d\tanh(x)}{dx} = \frac{d}{dx}\left[\frac{e^x - e^{-x}}{e^x + e^{-x}}\right] = \frac{(e^x + e^{-x})^2 - (e^x - e^{-x})^2}{(e^x + e^{-x})^2} = 1 - \left( \frac{e^x - e^{-x}}{e^x + e^{-x}} \right)^2 = 1 - \tanh^2(x)
$$
